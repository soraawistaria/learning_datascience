Read dataset

```python
import pandas as pd 

df = pd.read_csv('data_music.csv')

df.head()
```

```python
df.shape
df.isnull().sum()
```

```python
df = df[['danceability', 'energy', 'valence', 'tempo']]
df
```

Visualisasi

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")

sns.scatterplot(data=df, x='valence', y='danceability', s=100)
#s=100 artinya size dot pada grafik

plt.title("Korelasi Valence dengan Danceability pada Musik")
plt.xlabel("Valence")
plt.ylabel("Danceability")

plt.show()
```

Algoritma Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor
X = df[['tempo', 'energy', 'valence']]
Y = df['danceability']

model = RandomForestRegressor(random_state=42)
model.fit(X, Y)

print(f"Skor akurasi model : {model.score(X, Y):.4f}")
```
