Model Prediksi Nilai Ujian Siswa berdasarkan Jam Belajar Perminggu menggunakan algoritma linear regression

```python
import pandas as pd 

df = pd.read_csv('data_jamdannilai.csv', sep=',')

df.head(5)
```

```python
df.describe()
```

```python
df = df.drop_duplicates(subset=['Jam_Belajar'], keep='first')
df.head()
```

```python
df['Kategori_siswa'] = df['Jam_Belajar'].apply(lambda x: 'Rajin' if x >= 8 else 'Biasa')

df.head()
```

Data Visualizing

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")

sns.scatterplot(data=df, x='Jam_Belajar', y='Nilai_Ujian', hue='Kategori_siswa', s=100)
#s=100 artinya size dot pada grafik

plt.title("Hubungan Jam Belajar dengan Nilai")
plt.xlabel("Jam belajar per minggu")
plt.ylabel("Nilai akhir")

plt.show()
```

Machine Learning

```python
from sklearn.linear_model import LinearRegression
```

```python
X = df[['Jam_Belajar']]
y = df['Nilai_Ujian']

model = LinearRegression()
model.fit(X, y)

jam_bljr_Kaito = pd.DataFrame([[7]], columns=['Jam_Belajar'])
prediksi_nilai = model.predict(jam_bljr_Kaito)

print(f"Prediksi nilai Kaito = {prediksi_nilai}")
```
