Read dataset

```python
import pandas as pd 

df = pd.read_csv('train.csv')

df.head(5)
```

Exploratory Data Analysis (EDA)

```python
df.info()

df.isnull().sum()
```

```python
print("Group By Sex")
df.groupby('Sex')['Survived'].mean()
```

```python
print("Group By Pclass")
df.groupby('Pclass')['Survived'].mean()
```

Visualisasi

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.barplot(x='Sex', y='Survived', data=df, ax=axes[0], errorbar=None)
axes[0].set_title('Tingkat Keselamatan Berdasarkan Jenis Kelamin')

sns.barplot(x='Pclass', y='Survived', data=df, ax=axes[1], errorbar=None)
axes[1].set_title('Tingkat Keselamatan Berdasarkan Kelas Tiket')

plt.tight_layout()
plt.show()
```
