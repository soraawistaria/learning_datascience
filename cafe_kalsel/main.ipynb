Read dataset

```python
import pandas as pd 

df = pd.read_csv('cafe_scrap.csv')

df
```

```python
pd.set_option('display.max_info_columns', 50)

df.info()
```

```python
for i, col in enumerate(df.columns):
    print(f"{i}: {col}")
```

```python
df = df.drop(columns=['reviews_link', 'user_reviews_extended', 'emails', 'user_reviews', 'thumbnail', 'street_view_url'])

for i, col in enumerate(df.columns):
    print(f"{i}: {col}")
```

```python
df = df.drop_duplicates()

df.info()
```

```python
df = df.drop(columns=['descriptions', 'status', 'price_range', 'credit_cards_accepted', 'about'])

df.info()
```

```python


df = df.drop(columns=['phone', 'plus_code', 'website'])

df.info()
```

```python
df = df.drop(columns=['timezone', 'reservations', 'order_online', 'complete_address', 'owner', 'order_online'])

df.info()
```

```python
df = df.dropna(subset=['category'])

df = df.drop_duplicates(subset=['place_id'])

df = df.drop_duplicates(subset=['cid'])

df = df.drop_duplicates(subset=['data_id'])

df.info()
```

```python
print(df['category'].unique())
```

```python
df[df['category'] == 'Night club']
```

```python
kategori_dihapus = ['Night club', 'Car wash', 'Clothing store', 'Karaoke bar', 'Health consultant', 'Bar']

# Simpan baris yang kategorinya BUKAN bagian dari daftar di atas
df = df[~df['category'].isin(kategori_dihapus)]
df.info()
```

```python
# Simpan DataFrame ke file CSV
# index=False digunakan agar nomor baris pandas tidak ikut tersimpan sebagai kolom baru
df.to_csv('clean_data_cafekalsel.csv', index=False, encoding='utf-8')
```
