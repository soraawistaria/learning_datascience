Read Dataset

```python
import pandas as pd 

df_raw = pd.read_csv('cafe_kalsel.csv')
df_raw.head(5)
```

Ganti nama kolom

```python
df = df_raw[
    ['OSrXXb', 'rllt__details', 'rllt__details 2', 'ncRHuf', 'wA1Bge src']].rename(
    columns={
        'OSrXXb' : "nama_cafe",
        'rllt__details' : "kategori",
        'rllt__details 2' : "kota_prov",
        'ncRHuf' : "bisa_dinein",
        'wA1Bge src' : "link_gmaps"
    }
)

```

Memilah kategori

```python
df['kategori'] = (df['kategori'].astype(str).str.replace('· ', '').str.strip())

df = df[
    df['kategori'].str.contains('cafe|coffee|bakery', case=False, na=False)
]

```

Simpan yang kota nya jelas

```python
df = df[
    df['kota_prov'].str.contains('kalimantan|city', case=False, na=False)
]

pd.set_option('display.max_rows', None)

df
```
