```python
import pandas as pd 

df = pd.read_csv('tesdata1.csv', sep=';')

df.head(5)
```

```python
df.describe()
```

```python
df = df.drop_duplicates(subset=['nama'], keep='first')
df.head()
```

```python
df['Kategori_siswa'] = df['jam_bljr'].apply(lambda x: 'Rajin' if x >= 8 else 'Biasa')

df.head()
```

Data Visualizing

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")

sns.scatterplot(data=df, x='jam_bljr', y='nilai', hue='Kategori_siswa', s=100)

plt.title("Hubungan Jam Belajar dengan Nilai")
plt.xlabel("Jam belajar per hari")
plt.ylabel("Nilai akhir")

plt.show()
```

Machine Learning

```python
from sklearn.linear_model import LinearRegression
```

```python
X = df[['jam_bljr']]
y = df['nilai']

model = LinearRegression()
model.fit(X, y)

jam_bljr_Kaito = pd.DataFrame([[7]], columns=['jam_bljr'])
prediksi_nilai = model.predict(jam_bljr_Kaito)

print(f"Prediksi nilai Kaito = {prediksi_nilai}")
```
